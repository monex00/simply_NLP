```python
import pandas as pd
from collections import defaultdict, Counter
import nltk
from nltk.tokenize import word_tokenize
import numpy as np
```

```python
df = pd.read_csv('data/tweets.csv')
df.head()   
```

```python
def preprocessing(text):
    words = word_tokenize(text)
    filtered_words = []
    for word in words:
        print(word)
        if word == "&":
            filtered_words.append("and")
        if word == "amp":
            continue

        #puntuation
        if word in ['.', ',', '!', '?', ';', ':']:
            filtered_words.append(word)
        elif word.isalpha():
            filtered_words.append(word.lower())
        elif word.isalnum():
            filtered_words.append(word)
        
    return ['<s>'] + filtered_words + ['</s>']

preprocessing("I have always liked &amp; respected")
```

```python
df['tokens'] = df['text'].apply(preprocessing)
df.head()
```

```python
word_counts = Counter()
for tokens in df['tokens']:
    word_counts.update(tokens)

word_counts.most_common(10)
```

```python
for tokens in df['tokens'].iloc[:2]:
    print(tokens[:-1])
    print(tokens[1:])
    x = zip(tokens[:-1], tokens[1:])
    print(list(x))
    print()
```

```python
bigram_counts = defaultdict(Counter)
for tokens in df['tokens']:
    for token1, token2 in zip(tokens[:-1], tokens[1:]): 
        bigram_counts[token1].update([token2])

bigram_counts['<s>'].most_common(10)
```

```python
for tokens in df['tokens'].iloc[:2]:
    print(tokens[:-2])
    print(tokens[1:-1]) # skip the first token and the last token
    print(tokens[2:]) # skip the first two tokens
    x = zip(tokens[:-2], tokens[1:-1], tokens[2:])
    print(list(x))
    print()
```

```python
trigram_counts = defaultdict(Counter)
for tokens in df['tokens']:
    for token1, token2, token3 in zip(tokens[:-2], tokens[1:-1], tokens[2:]): 
        trigram_counts[(token1, token2)].update([token3]) # [token3] bc without [] it will update each character

trigram_counts[('<s>', 'Just')].most_common(10)
```

```python
probabilities = []
for candidate_word, count in bigram_counts["Just"].items():
    probabilities.append((candidate_word, count / word_counts["Just"]))

probabilities[:10]
```

```python
def bigram_model(init_seed = "<s>", finish_seed = "</s>", length = -1):
    current_word = init_seed
    sentence = []
    while (length == -1 or len(sentence) < length) and current_word != finish_seed:
        sentence.append(current_word)
        probabilities = []
        for candidate_word, count in bigram_counts[current_word].items():
            probabilities.append(count / word_counts[current_word])
        current_word = str(np.random.choice(list(bigram_counts[current_word].keys()), p=probabilities))
    return ' '.join(sentence)
bigram_model()
```

```python
def trigram_model(init_seed = ("<s>", ), finish_seed = "</s>", length = -1):
    current_words = init_seed
    sentence = [init_seed[0]]
    while (length == -1 or len(sentence) < length) and current_words[-1] != finish_seed:
        
        # if current words is a single word, we need to use bigram model to get the next word
        if(len(current_words) == 1):
            next_word = bigram_model(init_seed = current_words[0], finish_seed = finish_seed, length = 2)
            current_words = (current_words[0], next_word.split()[1])
            
        sentence.append(current_words[1])
        probabilities = []

        if current_words not in trigram_counts:
            break

        for candidate_word, count in trigram_counts[current_words].items():
            probabilities.append(count / bigram_counts[current_words[0]][current_words[1]])
        first_word, second_word = current_words
        next_word = np.random.choice(list(trigram_counts[(first_word, second_word)].keys()), p=probabilities)
        current_words = (second_word, str(next_word))
        
    return ' '.join(sentence)

trigram_model()
```
