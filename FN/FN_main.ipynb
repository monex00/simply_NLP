```python
andrea_frames = {'Touring': 1907, 'Cause_fluidic_motion': 920, 'Submitting_documents': 1521, 'Appellations': 2390, 'Evidence': 25}
fabio_frames = {'Strictness': 75, 'Being_pregnant': 2921, 'Sex': 2913, 'Wearing': 160, 'Dominate_situation': 1795}
```
